```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/tips.csv")
df.head()
```

```python
print("Размер таблицы:", df.shape)
```

```python
df.dtypes
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df["total_bill"], bins=15)
plt.title("Распределение суммы счета")
plt.xlabel("Сумма счета")
plt.ylabel("Количество наблюдений")
plt.show()
```

```python
day_counts = df["day"].value_counts()

plt.figure(figsize=(8, 5))
plt.bar(day_counts.index, day_counts.values)
plt.title("Количество посещений по дням недели")
plt.xlabel("День недели")
plt.ylabel("Количество посещений")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df["total_bill"], df["tip"])
plt.title("Связь суммы счета и чаевых")
plt.xlabel("Сумма счета")
plt.ylabel("Чаевые")
plt.show()
```

## Наблюдения

1. Наибольшее количество наблюдений приходится на субботу, а наименьшее — на пятницу.

2. В датасете представлены разные группы посетителей по полу, наличию курения, дню и времени посещения, поэтому данные можно использовать для сравнения этих категорий.

3. Между суммой счета и размером чаевых наблюдается положительная связь: при увеличении суммы счета чаевые в большинстве случаев также увеличиваются.
